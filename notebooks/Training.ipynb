```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
import pickle
```

```python
data = pd.read_csv("Churn_Modelling.csv")
data.head()
```

# Preprocess the data
### Drop irrelevant columns  - RowNumber CustomerID Surname

```python
data=data.drop(['RowNumber', 'CustomerId', 'Surname'],axis=1) # axis = 1 (Columns wise)
data.head()
```

# Encode Categorical Variables

```python
label_encoder = LabelEncoder()
data['Gender']=label_encoder.fit_transform(data['Gender'])
data.head(5)
```

# ONE HOT ENCODE 'GEOGRAPHY'

```python
from sklearn.preprocessing import OneHotEncoder
one_hot_encode = OneHotEncoder()

geo_encoder=one_hot_encode.fit_transform(data[['Geography']])

geo_encoder
```

```python
geo_encoder.toarray()
```

```python
one_hot_encode.get_feature_names_out(['Geography'])
```

```python
geo_encoded_df=pd.DataFrame(geo_encoder.toarray() ,columns=one_hot_encode.get_feature_names_out(['Geography']))
geo_encoded_df.head()
```

# Combinng Columns with the original Data

```python
data = pd.concat([data.drop('Geography',axis=1),geo_encoded_df],axis=1)
data.head(10)
```

## Save the encoders and Sscaler

```python
with open('label_encoder_gender.pkl','wb') as file:
    pickle.dump(label_encoder,file)

with open('onehot_encoder_geo.pkl','wb') as file:
    pickle.dump(one_hot_encode,file)
```

```python
data.head(10)
```

# Divide the dataset into independent and dependent features

```python
x = data.drop('Exited',axis=1)

y = data['Exited']

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2, random_state=42)

scaler = StandardScaler()

x_train = scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
with open('Scaler.pkl','wb') as file:
    pickle.dump(scaler,file)
```

# Artificial Neural Network Implementation
1 Sequential Network.  
2 Dense Layer - Defined How Much hidden layer.   
3 Activation functions - sigmoid, tanh, relu, leaky relu.  
4 Optimizer - Useful In Back propogation - Updating in the weights.  
5 Loss Function  
6 Metrics  
7 Logs - Training informstion into the folder -- Tensorbooard is used to display this logs like visualization graphs


```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping, TensorBoard
import datetime
```

```python
model = Sequential([
    Dense(64,activation='relu',input_shape=(x_train.shape[1],)), #First Hidden layer connected with the input layer
    Dense(32,activation='relu'), # HL2
    Dense(1,activation='sigmoid') # output layer
])
```

```python
model.summary()
```

# Compiling the model 

### Note 1 : As it is binary classification problem I am using binary_crossentropy here

### Note 3 : If it is multiclassification problem the sparse_crossentropy is used

```python
import tensorflow
opt = tensorflow.keras.optimizers.Adam(learning_rate=0.01)
loss=tensorflow.keras.losses.BinaryCrossentropy()
model.compile(optimizer=opt, loss=loss, metrics=['accuracy']) 

## This is the way we can initialise the optimizer and losses explicitly
# Advance - it allows us to set the our own learning rate in the optimizer.
```

```python
model.compile(optimizer="adam", loss="binary_crossentropy", metrics=['accuracy']) 
# This is how we can use optimizer & loss via inbuilt parameters but it does not allow us to set own learning rate.
```

```python
## Set up the Tensorboard
log_dir = "logs/fit/" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tensorflow_callback=TensorBoard(log_dir=log_dir,histogram_freq=1)
```

```python
## Set up Early Stopping
early_stopping_callback=EarlyStopping(monitor='val_loss', patience=20, restore_best_weights=True)
```

```python
#Train the model
history=model.fit(
    x_train,y_train, validation_data=(x_test, y_test), epochs=100, 
    callbacks=[tensorflow_callback, early_stopping_callback]
)
```

```python
# Save the model as h5 file beacuse it is compatible for keras
model.save('model.h5')
```
