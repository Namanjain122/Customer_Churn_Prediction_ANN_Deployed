# Load the Pickle File

```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import numpy as np
```

```python
model = load_model('model.h5')

## load the encoder and Scaler
with open('label_encoder_gender.pkl', 'rb') as file:
    label_encoder_gender=pickle.load(file)

with open('onehot_encoder_geo.pkl', 'rb') as file:
    onehot_encode_geo=pickle.load(file)

with open('Scaler.pkl', 'rb') as file:
    scalar=pickle.load(file)
```

```python
input_data = {
    'CreditScore' : int(input("enter your number")),
    'Geography' : input("enter your Country"),
    'Gender' : input("enter your gender"),
    'Age' : int(input("enter your age")),
    'Tenure' : int(input("enter Tenure")),
    'Balance' : int(input("enter your Balance")),
    'NumOfProducts' : int(input("enter your products")),
    'HasCrCard' : int(input("enter your HasCrCard")),
    'IsActiveNumber' : int(input("Enter your Active Number")),
    'EstimatedSalary' : int(input("enter your salary"))
}
```

```python
input_data = {
    'CreditScore' : 600,
    'Geography' : 'France',
    'Gender' : 'Male',
    'Age' : 40,
    'Tenure' : 3,
    'Balance' : 60000,
    'NumOfProducts' : 2,
    'HasCrCard' : 1,
    'IsActiveMember' : 1,
    'EstimatedSalary' : 50000
}
```

```python
geo_encoded=onehot_encode_geo.transform([[input_data['Geography']]]).toarray()
geo_encoded_df=pd.DataFrame(geo_encoded, columns=onehot_encode_geo.get_feature_names_out(['Geography']))
geo_encoded_df
```

```python
input_df=pd.DataFrame([input_data])
input_df
```

```python
input_df['Gender']=label_encoder_gender.transform(input_df['Gender'])
input_df
```

# Concatination & Prediction

```python
input_df=pd.concat([input_df.drop("Geography",axis=1), geo_encoded_df], axis=1)
input_df
```

```python
input_scaled = scalar.transform(input_df)
input_scaled
```

```python
prediction=model.predict(input_scaled)
prediction
```

```python
prediction_probab=prediction[0][0]
prediction_probab
```

```python
if prediction_probab>0.5:
    print("The Customer is likely to churn")
else:
    print("The customer is not likely to churn.")
```
